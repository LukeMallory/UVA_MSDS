```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns 



ames = pd.read_csv('data/ames_prices.csv')
ames["binary_air"] = (ames["Central.Air"] == "Y").astype(int)
```

a)

```python
proportion_ca = ames['binary_air'].mean()

print(f"Proportion of homes with central air: {proportion_ca}")
```

b)

```python
R = 2000
np.random.seed(5030)
results = []

for r in range(R):
    idx = np.random.randint(0, len(ames), len(ames))
    sample = ames['binary_air'].iloc[idx]
    prop = sample.mean()
    results.append(prop)

results = np.array(results)
results
```

c)

```python
standard_error = np.std(results, ddof=1)
print(f"Standard Error: {standard_error}")
```

d)

```python
cv_90 = 1.645
cv_95 = 1.96
cv_99 = 2.576

conf_levels = [cv_90, cv_95, cv_99]
for c in conf_levels:
    L = proportion_ca - c * standard_error
    R = proportion_ca + c * standard_error

    print(f"Confidence Interval [{L:.4f}, {R:.4f}]")
```

e)

* At α = 0.10, we reject the null hypothesis that the mean proportion is 0.9 because 0.9 is not present in the 90% confidence interval.

* At α = 0.05, we reject the null hypothesis that the mean proportion is 0.9 because 0.9 is not present in the 95% confidence interval.

* At α = 0.01, we reject the null hypothesis that the mean proportion is 0.9 because 0.9 is not present in the 99% confidence interval.

Because 0.9 is not present on the interval at all three levels of confidence, the intervals suggest that the proportion is not equal to 90.

# Problem 8:

```python
ames['arcsinh_area'] = np.arcsinh(ames['area'])
ames['arcsinh_price'] = np.arcsinh(ames['price'])

correlation = ames["arcsinh_area"].corr(ames["arcsinh_price"])
print(correlation)
```

a)

```python
R = 2000 #Repetitions
np.random.seed(5030) #seed
results = []

for r in range(R):
    idx = np.random.randint(0, len(ames), len(ames))

    sample_area = ames['arcsinh_area'].iloc[idx]
    sample_price = ames['arcsinh_price'].iloc[idx]

    corr = sample_area.corr(sample_price)
    results.append(corr)

results = np.array(results)
results
```

b)

```python
standard_error = np.std(results, ddof=1)
print(f"Standard Error: {standard_error}")
```

c)

```python
cv_90 = 1.645
cv_95 = 1.96
cv_99 = 2.576

conf_levels = [cv_90, cv_95, cv_99]
for c in conf_levels:
    L = correlation - c * standard_error
    R = correlation + c * standard_error

    print(f"Confidence Interval [{L:.4f}, {R:.4f}]")
```

d)

* At α = 0.10, we reject the null hypothesis that the correlation is 0 because 0 is not present in the 90% confidence interval.

* At α = 0.05, we reject the null hypothesis that the correlation is 0 because 0 is not present in the 95% confidence interval.

* At α = 0.01, we reject the null hypothesis that the correlation is 0 because 0 is not present in the 99% confidence interval.

The intervals suggest that the correlation coefficient is not equal to 0, because 0 is not present on any interval. Because the interval at the highest level of confident ranges from 0.6977 to 0.7486, there is evidence that area and price may have a strong positive correlation.

e)

Area and price must remain paired during the sampling because each sample represents an individual house. If the samples of area and price were drawn from different houses, the relationship would be irrelevant, and this would invalidate the sample.

# Problem 10

```python
nhanes = pd.read_csv("./data/nhanes_data_17_18.csv")

sleep = nhanes[(nhanes["AgeInYearsAtScreening"] >= 18) &
    (nhanes["SleepHoursWeekdaysOrWorkdays"].notna())
    ]["SleepHoursWeekdaysOrWorkdays"]


sleep_mean = sleep.mean()
print(sleep_mean)
```

a)

```python
R = 2000 #Repetitions
np.random.seed(5030) #Seed
results = []

for r in range(R):
    idx = np.random.randint(0, len(sleep), len(sleep))

    sample = sleep.iloc[idx]
    sample_mean = sample.mean()

    results.append(sample_mean)

results = np.array(results)
results
```

b)

```python
standard_error = np.std(results, ddof=1)
print(f"Standard Error: {standard_error}")
```

c)

```python
cv_90 = 1.645
cv_95 = 1.96
cv_99 = 2.576

conf_levels = [cv_90, cv_95, cv_99]

for c in conf_levels:
    L = sleep_mean - c * standard_error
    U = sleep_mean + c * standard_error

    print(f"Confidence Interval [{L:.4f}, {U:.4f}]")
```

d)

* At α = 0.10, we reject the null hypothesis that the mean is 8 because 8 is not present in the 90% confidence interval.

* At α = 0.05, we reject the null hypothesis that the mean is 8 because 8 is not present in the 95% confidence interval.

* At α = 0.01, we reject the null hypothesis that the mean is 8 because 8 is not present in the 99% confidence interval.

e)

At all three confidence intervals, there is statistically significant evidence that the average number of hours slept is not equal to 8. However, the practical significance of this test is limited because the predicted means are not significantly far from 8 hours in terms of how much the difference in sleep would affect an individual. Also, because the complex survey design is ignored, these results should not be considered nationally representative.
