3. Location 2 made an average of $10,769.60 less in sales than location 1.

```python
#4.
from scipy import stats
import numpy as np

rng = np.random.default_rng(1)
B = 5000

boot_spend_diff = np.empty(B)
for b in range(B):
    promotion1_star = rng.choice(
        sales1,
        size=len(sales1),
        replace=True
    )

    promotion2_star = rng.choice(
        sales2,
        size=len(sales2),
        replace=True
    )

    boot_spend_diff[b] = (
        promotion2_star.mean()
        - promotion1_star.mean()
    )

lower, upper = np.percentile(
    boot_spend_diff,
    [2.5, 97.5]
    )

print(
    f"95% bootstrap CI: "
    f"[${lower:.2f}, ${upper:.2f}]"
)
```

We are 95% coonfident that weekly sales with promotion 2 will be between $7,590 and $14,010 less than promotion 1.

If the confidence interval contains zero, this does not mean that there is no difference between the two promotions. It is possible that there is no difference, but this does not automatically guarantee that. The confidence interval is supposed to represent many values that the difference could plausibly be.
