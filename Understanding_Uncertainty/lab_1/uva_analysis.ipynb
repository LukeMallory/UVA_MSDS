```python
import pandas as pd
data = pd.read_csv("ForeignGifts_edu.csv")
data["Institution Name"].unique()
```

```python
uva = data[data['Institution Name'] == "University of Virginia"]
```

UVA Gifting Statistics

```python
print(f"Total amount recieved: ${uva['Foreign Gift Amount'].sum()}")
print(f"Average amount recieved: ${uva['Foreign Gift Amount'].mean()}")
print(f"Median amount recieved: ${uva['Foreign Gift Amount'].sum()}")
print(f"Number of gifts recieved: {uva['Foreign Gift Amount'].count()}")
```

UVA's Donors

```python
uva['Country of Giftor'].value_counts()
```

```python
top_giftors = uva.groupby('Country of Giftor')['Foreign Gift Amount'].sum()
top_giftors.sort_values()
```

```python
pd.reset_option('display.max_rows')
acc = data[data['Institution Name'].isin(["University of Virginia", 
           "University of North Carolina - Chapel Hill",
           'North Carolina State University',
           'Duke University',
           'University of Miami',
           'Clemson University',
           'Georgia Institute of Technology'
])]
acc["Institution Name"].value_counts()
```

```python
acc.groupby('Institution Name')['Foreign Gift Amount'].sum().sort_values()
```

UVA ranks third to last when in this sample of other ACC universities sorted by dollar amount of gifts recieved.

UVA ranks third to last in this sample group when sorted by average gift size.
